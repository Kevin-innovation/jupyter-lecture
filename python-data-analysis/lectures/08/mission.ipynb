- [ ] 선 그래프, 막대 그래프, 히스토그램, 산점도, 박스플롯, 히트맵 중 5종 이상을 사용했다.
- [ ] 각 차트에 제목과 축 라벨을 붙였다.
- [ ] 결론이 코드 출력 숫자와 모순되지 않는다.

## 질문 준비 메모

| 구간 | 스스로 확인할 질문 |
|---|---|
| 문제 1~2 | 날짜형 변환이 필요한 이유를 설명할 수 있는가? |
| 문제 3~6 | 선 그래프와 막대 그래프를 언제 다르게 쓰는가? |
| 문제 7~9 | 분포, 관계, 그룹별 퍼짐을 각각 어떤 차트로 보는가? |
| 문제 10~12 | 상관관계와 피벗 히트맵은 어떤 질문에 답하는가? |
| 문제 13~15 | 차트 제목과 결론이 실제 숫자와 연결되어 있는가? |

## 제출 형태 안내

- 차트만 제출하지 말고 각 차트가 답하는 질문을 한 줄로 적는다.
- 숫자 열을 비교할 때는 합계, 평균, 중앙값 중 무엇을 썼는지 표시한다.
- `plt.close(fig)` 는 제출 노트북에서 필수는 아니지만, 여러 차트를 만들 때 화면이 과하게 길어지는 것을 줄일 수 있다.
- 결론에는 가장 높은 항목, 비교 기준, 다음 분석 행동이 들어가야 한다.

## 풀이 전 작업 순서

1. 원본 CSV를 읽고 행 수, 열 이름, 결측 여부를 확인한다.
2. 날짜형 변환과 월 열 생성을 먼저 끝낸다.
3. 일별, 채널별, 카테고리별 요약표를 만든다.
4. 요약표를 기준으로 선 그래프와 막대 그래프를 만든다.
5. 원본 행 단위 데이터로 히스토그램, 산점도, 박스플롯을 만든다.
6. 상관계수와 피벗 테이블은 표를 먼저 확인한 뒤 히트맵으로 표현한다.
7. 여러 차트를 한 화면에 묶을 때는 제목과 축 라벨이 겹치지 않는지 확인한다.
8. 마지막 결론은 차트에서 본 숫자를 그대로 사용한다.

## 자주 나는 오류와 확인 방법

- `KeyError`: 열 이름 오타가 있는지 `df.columns` 를 확인한다.
- 날짜 그래프가 이상함: `date` 가 `datetime` 으로 바뀌었는지 확인한다.
- 막대 그래프 순서가 어색함: 먼저 요약표를 정렬하고 그 표로 그린다.
- 히트맵이 비어 있음: 숫자 열만 선택했는지 확인한다.
- 상관관계를 인과관계로 설명함: "같이 움직인다"와 "원인이다"를 구분한다.
- 결론과 차트가 맞지 않음: 결론을 쓰기 전 마지막 요약표를 다시 출력한다.

## 좋은 결론 조건

좋은 시각화 결론은 차트 이름을 나열하지 않는다. 어떤 기간, 채널, 카테고리에서 차이가 났는지 말하고, 다음 행동을 제안한다.

- 매출이 가장 높은 월 또는 채널
- 전환율이나 ROI가 높은 그룹
- 광고비와 매출의 관계
- 다음 분석에서 더 확인할 기준

예를 들어 "막대 그래프를 만들었다"는 부족하다. "web 채널의 매출이 가장 높지만 app 채널의 ROI가 더 높다면, 다음 캠페인은 매출 규모와 효율을 나누어 비교해야 한다"처럼 차트에서 행동으로 이어져야 한다.

## 차트 해석 체크리스트

시각화 과제는 코드가 실행되는 것만으로 끝나지 않는다. 제출 전에 아래 항목을 보고 각 차트가 분석 질문에 답하는지 확인한다.

| 차트 | 확인할 질문 |
|---|---|
| 선 그래프 | 시간 순서가 있고 증가·감소 흐름을 말할 수 있는가? |
| 막대 그래프 | 범주별 크기 비교가 명확한가? |
| 히스토그램 | 값이 몰린 구간과 퍼진 구간을 설명할 수 있는가? |
| 산점도 | 두 숫자가 함께 움직이는지, 예외가 있는지 보이는가? |
| 박스플롯 | 그룹별 중앙값과 퍼짐 차이를 말할 수 있는가? |
| 히트맵 | 색으로 본 차이를 실제 숫자로 다시 확인했는가? |

차트가 많아도 같은 질문만 반복하면 좋은 리포트가 아니다. 최종 제출에서는 시간 흐름 1개, 범주 비교 1개, 분포 1개, 관계 1개, 요약 히트맵 1개처럼 서로 다른 질문을 다루는 차트를 남긴다.
