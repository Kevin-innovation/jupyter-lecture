## 미션 시나리오

당신은 **"헬스리포트 주식회사"** 신입 데이터 분석가다. 회사가 사내 직원 한 명의 90일 활동 데이터를 받았다(걸음 수). 같은 기간의 외부 기온 데이터도 함께 받았다. 이 정보를 정리해 마케팅팀에 다음 질문에 답하는 리포트를 제출해야 한다.

> **마케팅 팀장의 질문 4가지**
>
> 1. 이 직원은 한 달 평균과 두/세 번째 달 평균이 어떻게 다른가? 점점 활동이 늘었나, 줄었나?
> 2. 추운 날에는 덜 걷는 경향이 있는가? 데이터로 입증할 수 있나?
> 3. 평일과 주말 중 어느 쪽이 활동이 부족한가?
> 4. 다음 분기에 이 사람을 위한 캠페인 메시지를 만든다면 어떤 톤으로 갈까?

이 질문에 답하는 분석 노트북을 작성한다.

3. **월별 비교**: 1~30일, 31~60일, 61~90일을 각각 한 달로 보고 각각의 평균 걸음을 구해 비교한다.
4. **기온 분기 비교**: 기온이 평균 미만인 날 vs 평균 이상인 날의 평균 걸음을 비교한다.
5. **평일/주말 비교**: `(day-1) % 7` 이 5 또는 6 인 날을 주말로 보고, 평일/주말 평균 걸음을 비교한다.
6. 마지막 마크다운 셀에 **결론 4줄**을 작성한다. 팀장 질문 4개에 각각 한 줄씩.

## 보너스 요구사항 (선택, 가산점)

> **🥄 수학 지식 한스푼 — 상관계수 (correlation coefficient, r)**
>
> - **뜻**: 두 변수가 같이 움직이는 정도를 −1 ~ +1 사이의 한 숫자로 요약한 값.
> - **읽는 법**: **+1 에 가까울수록** 한쪽이 커질 때 다른 쪽도 같이 커짐 / **−1 에 가까울수록** 한쪽이 커질 때 다른 쪽은 작아짐 / **0 에 가까우면** 관련 없음.
> - **대략 기준**: |r| < 0.3 약함, 0.3~0.7 보통, > 0.7 강함.
> - **주의**: 상관관계 ≠ 인과관계. 두 값이 같이 움직인다고 한쪽이 다른 쪽의 원인은 아니다 (예: 아이스크림 매출과 익사 사고 수는 양의 상관인데 둘 다 "여름" 이라는 공통 원인 때문).
> - **NumPy 호출**: `np.corrcoef(a, b)` 는 2×2 행렬을 돌려준다. 우리가 보고 싶은 값은 `[0, 1]` 위치.

다음 중 하나 이상 시도하면 가산점.

- ⭐ **B1 — 상관계수**: `np.corrcoef(temp, steps)[0, 1]` 로 기온과 걸음 수의 상관관계를 계산하고 부호와 크기를 위 기준으로 해석한다.
- ⭐ **B2 — 최악의 주**: 90일을 7일씩 끊어 "한 주 평균" 13개(마지막은 6일)를 만들고, 가장 부족했던 주가 몇 주차였는지 출력한다.
- ⭐ **B3 — 10,000보 연속 달성**: 10,000보 이상 걸은 날이 연속으로 몇 일까지 이어졌는지 최대 길이를 출력한다 (NumPy 만으로 풀이 가능, 어려움).
