## 1. 데이터 분석이란 무엇인가

데이터 분석은 **숫자에 깔린 이야기를 꺼내는 일**이다. 처음 듣는 학생을 위해 한 줄로 정리하면 다음과 같다.

> 데이터 분석 = 데이터를 모아 → 깨끗하게 만들고 → 패턴을 찾아 → 누군가에게 보여주는 것.

이 한 줄을 잡지식·역사·현업 직무와 함께 풀어 본다.

### 1-1. 한 줄 역사 메모

분야의 이름은 통계학자 **John Tukey** 가 1977년 책 *Exploratory Data Analysis* (EDA) 로 만들었다. 그의 한마디: *"숫자를 보기 전에 가설부터 세우지 말라. 데이터가 말해줄 때까지 그림과 표로 들여다봐라."* — 이 코스가 가르치는 게 정확히 그 일이다.

### 1-2. 분석 파이프라인 4단계

어느 직무든 작업의 큰 흐름은 같다.

```
1. 수집(Collect)     : 어디에서 데이터를 가져올 것인가
2. 정제(Clean)       : 빠진 값, 잘못된 값, 중복 처리
3. 탐색(Explore)     : 통계, 그룹화, 시각화로 패턴 찾기
4. 전달(Communicate) : 결론을 표/차트/한 문장으로 정리
```

업계 클리셰: 분석 시간의 70~80% 가 **2번 정제**에 쓰인다(체감은 05번 레슨에서). 1번 레슨은 4단계 중 **3번 탐색** 만 다룬다.

### 1-3. 분석 vs 머신러닝

**분석은 이미 일어난 일을 설명**하고("지난달 매출은 X 였다"), **머신러닝은 앞으로 일어날 일을 예측**한다("다음 달 매출은 Z 일 것이다"). 머신러닝도 좋은 분석 위에서만 잘 동작한다.

### 1-4. 왜 NumPy 부터 시작하는가

데이터 분석의 거의 모든 라이브러리(`pandas`, `scikit-learn`, `tensorflow`, `pytorch`) 가 NumPy 위에서 동작한다. NumPy 는 "숫자를 빠르게 다루는 표준 그릇" 이라 가장 먼저 익혀야 한다. **파이썬 리스트는 객체 포인터 모음으로 유연하지만 느리고**, **NumPy `ndarray` 는 같은 타입 숫자가 메모리에 연속해서 박혀 있어 빠르고 메모리도 적게 쓴다.** 이 차이는 6절에서 직접 측정한다.

## 3. NumPy 입문 — ndarray 만들기

### 3-1. 한 줄 역사 메모

1995년 Numeric → 2001년 Numarray → **2005년 Travis Oliphant 가 둘을 통합해 NumPy 를 만들었다**. 같은 인물이 pandas 전신과 Anaconda 도 만들었다.

### 3-2. 왜 NumPy 가 빠른가

파이썬 리스트는 객체 포인터를 늘어놓아 메모리에 숫자가 흩뿌려진다. CPU 는 매번 포인터를 따라 한 칸씩 더한다. NumPy 의 `ndarray` 는 같은 타입 숫자가 메모리에 한 줄로 붙어 있고, C 로 쓰여 있어 CPU 의 **SIMD** (Single Instruction Multiple Data) 명령으로 여러 숫자를 동시에 더한다. 보통 **10~100배** 차이. 6절에서 직접 확인한다.

### 3-3. 파이썬 리스트에서 배열 만들기

````python
import numpy as np
scores = np.array([88, 92, 75, 60, 95])
print(scores, type(scores))
print("shape:", scores.shape, "| dtype:", scores.dtype, "| size:", scores.size, "| ndim:", scores.ndim)
````

`shape` 는 각 차원의 길이를 담은 튜플로 **가장 자주 본다**. `(5,)` 는 "원소 5개인 1차원" (콤마 뒤가 비어 있음). `(5, 1)` 은 2차원, `(5, 1, 1)` 은 3차원. `dtype` 은 원소 자료형(보통 `int64` 또는 `float64`), `size` 는 전체 원소 개수(shape 곱), `ndim` 은 차원 수.

### 3-4. CSV 에서 배열로 가져오기

`data/scores.csv` 에 **200명** 의 점수가 있다.

````python
scores = np.loadtxt(f"{DATA_BASE}/scores.csv", delimiter=",", skiprows=1)
print("처음 10명:", scores[:10])
print("총 학생 수:", scores.shape[0], "| dtype:", scores.dtype)
````

`delimiter=","` = 구분자, `skiprows=1` = 헤더 건너뛰기. `skiprows` 없으면 헤더 문자 때문에 `ValueError: could not convert string to float: 'score'`. `np.loadtxt` 는 항상 `float64` 로 읽으니 정수 필요 시 `.astype(int)`.

### 3-6. dtype 깊이 — 정수와 실수의 종류

NumPy 의 dtype 은 메모리를 얼마나 차지하느냐로 갈린다. 정수는 `int8`(±127) / `int16`(±32K) / `int32`(±21억) / `int64`(±9×10^18), 실수는 `float32`(유효숫자 ≈7자리) / `float64`(≈15자리). 기본은 `int64`·`float64`. 큰 데이터에서 메모리를 아낄 때만 작은 타입을 명시한다.

잡지식: `float64` 는 **IEEE 754** "double precision" 형식이다. `0.1 + 0.2 == 0.3` 이 `False` 인 그 형식. 금융 시스템은 이래서 `decimal.Decimal` 이나 정수 센트 단위로 다룬다.

````python
print(0.1 + 0.2)        # 0.30000000000000004
print(0.1 + 0.2 == 0.3) # False
````

````python
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# [문제 2]  기술통계 함수 8가지 출력하기
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
#
# 💡 힌트
#   scores.size      →  원소 개수
#   scores.sum()     →  합계
#   scores.mean()    →  평균
#   scores.std()     →  표준편차
#   scores.var()     →  분산
#   np.median(scores)→  중앙값
#   scores.min()     →  최솟값
#   scores.max()     →  최댓값
#

````

### 4-1. 평균과 중앙값의 차이

> **🥄 수학 지식 한스푼 — 평균(mean, x̄) / 중앙값(median) / 최빈값(mode)**
>
> - **뜻**: 데이터를 한 숫자로 줄였을 때 "대표값" 세 가지.
>   - 평균: 모두 더하고 개수로 나눔.
>   - 중앙값: 작은 순서로 정렬했을 때 한가운데 값.
>   - 최빈값: 가장 자주 나오는 값.
> - **수식**: 평균 `x̄ = (x₁ + x₂ + … + xₙ) / N`
> - **읽는 법**: 평균은 모든 값을 골고루 반영하는 대신 극단값에 끌려간다. 중앙값은 끌려가지 않는다.
> - **예시**: 4명 점수 `[60, 70, 80, 200]` → 평균 102.5, 중앙값 75. 평균 102 점을 받은 사람은 아무도 없다.

위 박스의 핵심 한 줄: **평균은 극단값에 민감, 중앙값은 둔감**. 점수 데이터(이번 미션 1) 처럼 분포가 좌우로 비슷하면 둘이 거의 같다. 가격 데이터(미션 4) 처럼 한쪽으로 치우치면 둘이 다르게 나온다.

> 평균과 중앙값이 크게 다르면 분포가 한쪽으로 치우쳤다는 신호다. **평균 > 중앙값** 은 오른쪽 꼬리(큰 값 쪽) 가 긴 분포 — 소득·가격이 대표. **평균 < 중앙값** 은 그 반대.

이 직관을 미션 4 의 가격 데이터에서 직접 본다.

### 4-2. 표준편차와 분산

> **🥄 수학 지식 한스푼 — 분산(variance, σ²) / 표준편차(standard deviation, σ)**
>
> - **뜻**: 데이터가 평균에서 얼마나 흩어져 있는지를 숫자 하나로 요약한 것.
> - **수식**: 분산 `σ² = Σ(xᵢ − x̄)² / N`, 표준편차 `σ = √σ²`
> - **만드는 순서**: ① 각 값에서 평균을 뺀다 → ② 그 차이를 제곱한다(부호 없애기 + 큰 차이 강조) → ③ 모두 더하고 개수로 나눈다(=분산) → ④ 제곱근을 씌운다(=표준편차).
> - **읽는 법**: 클수록 넓게 퍼짐, 작을수록 평균 근처에 몰림. 표준편차는 단위가 원본과 같아(점수면 점) 해석이 쉽다.
> - **예시**: 두 반 평균이 똑같이 70점인데 한 반은 σ=5, 다른 반은 σ=20. 첫 반은 거의 모두 65~75점, 둘째 반은 50점과 90점이 섞여 있다.

> **🥄 수학 지식 한스푼 — 자유도(degrees of freedom) / `ddof`**
>
> - **뜻**: 분산을 구할 때 분모로 N 을 쓰느냐, N−1 을 쓰느냐를 결정하는 개념.
> - **두 가지**: **모분산**은 데이터 전체를 다 가졌을 때 쓰는 값(N 으로 나눔). **표본분산**은 일부 표본만 갖고 전체를 추측할 때 쓰는 값(N−1 로 나눔, 살짝 더 크게 나옴).
> - **왜 N−1?**: 표본에서는 평균을 추정해서 쓰는데, 그 과정에서 정보 한 개를 "써 버렸다" 고 본다. 그래서 분모에서 1을 뺀다.
> - **NumPy 기본**: `ddof=0` (모분산). **pandas 기본**: `ddof=1` (표본분산). 결과가 살짝 다르면 거의 항상 이 차이다.
> - **예시**: 50명 점수 데이터가 "전국 학생 추정용 표본"이면 `ddof=1`, "그 반 전체" 면 `ddof=0`.

1번 레슨에서는 NumPy 기본(모분산)을 그대로 쓴다.

### 4-3. 분위수 (quantile)

> **🥄 수학 지식 한스푼 — 분위수(quantile) / 사분위수(quartile) / IQR**
>
> - **뜻**: 데이터를 작은 순서로 정렬한 다음 일정 비율 지점에서 끊는 값.
>   - **분위수(quantile)**: 비율을 자유롭게 잡음. 예: 10% 분위수, 73% 분위수.
>   - **사분위수(quartile)**: 4등분 지점. Q1=25%, Q2=50%(=중앙값), Q3=75%.
>   - **IQR (Interquartile Range)**: Q3 − Q1. 데이터의 가운데 50% 가 차지하는 폭.
> - **읽는 법**: Q3 가 100 이고 Q1 이 80 이면 "가운데 50% 학생이 80~100점에 있다" 는 뜻. IQR 이 크면 가운데도 넓게 퍼짐, 작으면 가운데가 좁게 몰림.
> - **이상치 판정에 활용**: 보통 `Q1 − 1.5·IQR` 보다 작거나 `Q3 + 1.5·IQR` 보다 크면 이상치로 본다. 08번 boxplot 의 점이 이걸 시각화한 것.
> - **예시**: 200명 시험에서 Q3=85 면 "상위 25% 컷이 85점" 이라는 한 줄로 학원 상담에서 바로 쓰인다.

````python
q25, q50, q75 = np.quantile(scores, [0.25, 0.50, 0.75])
iqr = q75 - q25
print(f"Q1: {q25}, Q2: {q50}, Q3: {q75}, IQR: {iqr}")
````

(잡지식: `np.quantile(..., method=...)` 로 9가지 보간법을 고를 수 있고 기본은 `'linear'` 다.)

### 4-4. argmax 와 argmin

> **🥄 수학 지식 한스푼 — argmax / argmin**
>
> - **뜻**: 최댓값(또는 최솟값) **그 자체** 가 아니라, 그 값을 만들어낸 **위치(인덱스)** 를 찾아주는 함수.
> - **이름의 뜻**: "arg" 는 "argument(인수)" 의 줄임. 영문 그대로 풀면 "max 를 만든 인수가 무엇인가" 라는 질문.
> - **읽는 법**: `arr.max()` 가 "1등의 점수" 를 알려준다면, `arr.argmax()` 는 "1등이 몇 번째 학생인지" 를 알려준다.
> - **예시**: 90일 걸음 데이터에서 `steps.argmax()` 는 "가장 많이 걸은 날이 며칠째인지" 를 알려준다. 머신러닝에서도 "예측 확률이 가장 높은 클래스를 고른다" 가 정확히 argmax 다.

````python
i_max, i_min = scores.argmax(), scores.argmin()
print(f"최고점: {scores[i_max]} (인덱스 {i_max}번)")
print(f"최저점: {scores[i_min]} (인덱스 {i_min}번)")
````

## 5. 인덱싱과 슬라이싱

배열의 일부만 꺼내는 방법이다.

### 5-1. 기본 인덱싱

````python
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# [문제 3]  인덱싱·슬라이싱 6가지 패턴
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
#
# 💡 힌트
#   scores[0]    →  첫 번째 원소
#   scores[-1]   →  마지막 원소
#   scores[:5]   →  처음 5개
#   scores[-5:]  →  마지막 5개
#   scores[::2]  →  짝수 인덱스만
#   scores[::-1] →  역순 전체
#

````

파이썬 리스트와 문법이 같다. 음수 인덱스, 슬라이스 모두 그대로 동작한다.

### 5-2. view vs copy 잡지식

파이썬 리스트와 달리 **NumPy 슬라이스는 view(원본 공유)** 다. 메모리 효율을 위한 결정이다. 슬라이스를 수정하면 원본도 바뀐다.

````python
a = np.array([1, 2, 3, 4, 5])
b = a[:3]
b[0] = 999
print(a)   # [999, 2, 3, 4, 5] — 원본도 바뀜!
````

독립 사본이 필요하면 `.copy()` 를 명시한다. 큰 분석에서 이 차이를 모르고 짜면 디버깅이 어려워진다.

### 5-3. 조건으로 거르기 — boolean indexing

여기가 NumPy 의 진짜 위력이다. 파이썬 `for` 없이 한 줄로 필터링한다.

````python
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# [문제 4]  boolean indexing으로 점수 분류하기
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
#
# 💡 힌트
#   high = scores[scores >= 90]   →  90점 이상 추출
#   low  = scores[scores < 60]    →  60점 미만 추출
#   print('90점 이상:', high)
#   print('60점 미만:', low)
#   print(f'90점 이상 {high.size}명, 60점 미만 {low.size}명')
#

````

작동 원리는 두 단계다.

````python
mask = scores >= 90        # [False, False, True, ...] 같은 불리언 배열
print(mask[:10])
filtered = scores[mask]    # True 위치의 값만 꺼냄
print(filtered[:5])
````

`scores >= 90` 자체가 `True/False` 가 늘어선 배열이고, 배열에 그걸 그대로 넣으면 True 자리의 값만 꺼낸다. 길이가 같은 마스크여야 한다.

여러 조건은 `&` (and), `|` (or), `~` (not) 으로 조합한다. **`and`, `or` 는 안 된다.** 비트 연산자만 동작한다.

````python
mid = scores[(scores >= 60) & (scores < 90)]   # 60 이상 90 미만
extreme = scores[(scores < 50) | (scores >= 95)] # 극단값
print(f"중간대: {mid.size}명, 극단: {extreme.size}명")
````

`&`, `|` 는 우선순위가 비교 연산자보다 높으므로 **반드시 괄호로 묶어야** 한다. 안 묶으면 조용히 잘못된 결과가 나온다.

## 6. 벡터화 vs for 루프 — 직접 측정

NumPy 가 빠르다는 말의 실체를 측정으로 확인한다.

````python
import time
big = np.arange(1, 10_000_001)   # 1천만 개 정수

t0 = time.time()
total = 0
for x in big: total += x
t_for = time.time() - t0

t0 = time.time()
total_np = big.sum()
t_np = time.time() - t0

print(f"for 루프: {t_for:.3f}초")
print(f"NumPy   : {t_np:.4f}초")
print(f"속도 차이: 약 {t_for / t_np:.0f}배")
````

코랩에서 보통 30~80배 차이가 난다. 분석가는 어지간하면 for 루프를 쓰지 않고 **벡터화(vectorize)** — for 루프 대신 배열 단위 연산으로 — 한다.

## 8. 따라하기 — 90일 걸음 수 종합 분석

`data/daily_steps.csv` 에는 90일간 걸음 수가 들어 있다. 위에서 배운 모든 것을 한 번에 적용해 보자.

````python
import numpy as np

raw = np.loadtxt(f"{DATA_BASE}/daily_steps.csv", delimiter=",", skiprows=1)
day = raw[:, 0].astype(int)
steps = raw[:, 1]

print("관측 일수:", steps.size)
print("총 걸음  :", int(steps.sum()))
print(f"평균     : {steps.mean():.0f} 보 / 일")
print(f"중앙값   : {np.median(steps):.0f} 보 / 일")
print(f"표준편차 : {steps.std():.0f}")
print(f"최댓값   : {int(steps.max())} 보 (일자: {int(day[steps.argmax()])})")
print(f"최솟값   : {int(steps.min())} 보 (일자: {int(day[steps.argmin()])})")
````

````python
# 분위수
q25, q75 = np.quantile(steps, [0.25, 0.75])
iqr = q75 - q25
print(f"Q1: {q25:.0f}, Q3: {q75:.0f}, IQR: {iqr:.0f}")
````

````python
# 목표 달성 분석
goal = 10000
achieved_mask = steps >= goal
print(f"10,000 보 달성: {achieved_mask.sum()}일 / {steps.size}일")
print(f"달성률: {achieved_mask.mean() * 100:.1f} %")

# 잡지식: True/False 배열에 mean() 을 쓰면 True 의 비율이 나온다.
# True=1, False=0 으로 더해진 다음 개수로 나누기 때문.
````
