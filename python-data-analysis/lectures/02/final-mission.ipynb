## 보너스 요구사항 (선택, 가산점)

> **🥄 수학 지식 한스푼 — 샤프 지수 (Sharpe Ratio)**
>
> - **뜻**: 리스크 한 단위당 초과 수익률. 높을수록 "위험 대비 효율이 좋다"는 의미.
> - **수식**: Sharpe = (연환산 수익률 − 무위험 수익률) / 연환산 변동성. 여기서는 무위험 수익률 = 0 으로 단순화한다.
> - **읽는 법**: Sharpe > 1 이면 좋음, > 2 이면 우수, < 0 이면 무위험 자산만 못함.
> - **예시**: 워런 버핏 버크셔 해서웨이의 장기 Sharpe 는 약 0.76, S&P500 은 약 0.4.

- ⭐ **B1 — 샤프 지수**: 각 종목의 `ann_ret / ann_vol` 을 계산하고 가장 효율적인 종목을 찾는다.
- ⭐ **B2 — 상관 행렬**: `np.corrcoef(log_ret.T)` 로 5×5 상관 행렬을 계산하고 가장 상관이 높은 종목 쌍을 출력한다 (자기 자신 제외).
- ⭐ **B3 — 드로우다운**: 각 종목의 252일 중 누적 최고가 대비 하락률(drawdown)의 최댓값(MDD)을 계산한다.
