## 좋은 결론과 나쁜 결론

- 나쁜 결론: “분석 결과가 나왔다.” 어떤 숫자도 비교도 없다.
- 보통 결론: “A가 가장 높았다.” 기준과 수치가 부족하다.
- 좋은 결론: “A 기준 합계가 B보다 18% 높아, 다음 의사결정에서는 A를 우선 검토해야 한다.” 숫자, 비교, 행동이 모두 있다.

## 선택 보너스 안내

필수 요구사항을 끝낸 학생은 같은 데이터로 기준 하나를 더 바꿔 본다. 예를 들어 날짜, 지역, 채널, 학년, 등급, 프로그램 유형처럼 이미 존재하는 범주형 열을 하나 골라 비교하면 된다. 보너스는 새로운 어려운 함수를 쓰는 것이 아니라, 같은 사고방식을 다른 기준에 적용하는 연습이다.

제출물에는 보너스와 필수를 구분해 표시한다. 그래야 채점자가 필수 요구사항 누락 여부를 빠르게 확인할 수 있다.
