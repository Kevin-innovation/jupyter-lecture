- [ ] 정제 기준을 주석이나 결론에 남겼다.
- [ ] 저장 파일은 원본이 아니라 정제 결과다.
- [ ] 결론이 코드 출력 숫자와 모순되지 않는다.

| 문제 5~6 | 범주형 결측과 숫자형 결측을 서로 다르게 처리한 이유는 무엇인가? |

## 제출 형태 안내

- 원본 데이터는 `df`, 정제 데이터는 `clean` 으로 유지한다. 원본을 바로 덮어쓰지 않는다.
- 결측치와 이상치 처리는 정답 하나만 있는 것이 아니다. 대신 어떤 기준을 선택했는지 설명할 수 있어야 한다.
- 숫자형 결측은 중앙값, 범주형 결측은 `"U"` 또는 `"none"` 으로 통일하는 것을 기본 기준으로 둔다.
- 최종 결론에는 제거한 중복 수, 결측 처리 결과, 이상치 보정 기준, 다음 분석 방향을 포함한다.
- 오류가 나면 먼저 `clean.dtypes`, `clean.isna().sum()`, `clean.head()` 를 확인한다.

## 풀이 전 작업 순서

정제 문제는 순서가 중요하다. 아래 흐름을 지키면 정제 전후 비교가 명확해진다.

1. 원본 `df` 를 읽고 크기, 열 이름, 타입을 기록한다.
2. 결측치 개수와 결측 비율을 계산한다.
3. 중복 개수를 확인한 뒤 `clean` 을 만들어 중복을 제거한다.
4. 날짜와 숫자형 열을 분석 가능한 타입으로 변환한다.
5. 범주형 결측과 알 수 없는 값을 기준에 맞게 통일한다.
6. 숫자형 결측은 중앙값으로 채운다.
7. 이상치 개수와 예시를 먼저 확인한다.
8. 정해진 범위로 이상치를 보정한다.
9. 정제 전 이슈를 플래그로 남기고 품질 점수를 만든다.
10. 정제 후 결측과 범위가 모두 정상인지 검증한다.
11. 학년별·동아리별 요약을 만들고 최종 정제 파일을 저장한다.

## 제출 품질 기준

| 항목 | 좋은 제출 | 보완이 필요한 제출 |
|---|---|---|
| 원본 보존 | `df` 와 `clean` 을 분리 | 원본을 바로 덮어씀 |
| 결측 처리 | 숫자형/범주형 기준이 다름 | 모든 결측을 0으로 채움 |
| 이상치 처리 | 처리 전 개수와 처리 후 범위를 모두 확인 | 바로 clip 후 끝냄 |
| 품질 점수 | 정제 전 이슈를 기준으로 계산 | 정제 후 값만 보고 모두 정상 처리 |
| 저장 | 정제 결과를 새 CSV로 저장 | 원본 CSV 이름으로 저장 |
| 결론 | 중복, 결측, 이상치, 다음 행동 포함 | 숫자 없이 "정리했다"로 끝남 |

## 자주 나는 오류와 확인 방법

- `KeyError`: 열 이름 오타가 있는지 `df.columns` 를 확인한다.
- 날짜 변환 오류: `errors="coerce"` 를 사용하고 변환 실패 개수를 확인한다.
- 숫자형 계산 오류: `pd.to_numeric` 으로 타입을 먼저 바꾼다.
- 결측이 계속 남음: `value_counts(dropna=False)` 와 `isna().sum()` 을 함께 확인한다.
- 품질 점수가 모두 100점: 정제 후 값이 아니라 정제 전 이슈 기준으로 플래그를 만들었는지 확인한다.
- 저장 후 shape가 달라짐: `index=False` 를 사용했는지 확인한다.

## 결론 작성 가이드

최종 결론은 다음 네 요소를 포함한다.

- 원본 행 수와 정제 후 행 수
- 제거한 중복 행 수
- 결측 셀 수가 어떻게 바뀌었는지
- 이상치 보정 기준과 다음 분석 방향

예를 들어 "결측치를 처리했다"는 부족하다. "원본 결측 셀 244개를 범주형 U/none, 숫자형 중앙값 기준으로 처리해 정제 후 결측 0개로 만들었다"처럼 기준과 숫자가 함께 있어야 한다.
