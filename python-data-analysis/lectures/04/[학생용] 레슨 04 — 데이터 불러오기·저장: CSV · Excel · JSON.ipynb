## 제출 형태 안내

- 파일 경로는 `DATA_BASE` 와 `OUTPUT_PATH` 를 사용한다. 개인 컴퓨터 절대경로를 쓰지 않는다.
- 중간 표를 만들 때는 `sales`, `master`, `merged`, `events`, `final_report` 처럼 의미 있는 변수명을 사용한다.
- 저장 파일은 실행할 때마다 덮어써도 괜찮은 요약 결과만 저장한다.
- 결론에는 분기 번호와 숫자를 함께 쓴다. "4분기가 높다"보다 "4분기 매출이 가장 높다"처럼 기준을 밝혀 쓴다.
- 오류가 나면 먼저 `df.columns` 와 `df.shape` 를 출력해 지금 다루는 표가 의도한 표인지 확인한다.

## 풀이 전 작업 순서

이번 레슨은 파일이 여러 개라서 순서를 놓치기 쉽다. 아래 순서를 지키면 오류를 줄일 수 있다.

1. CSV 하나를 먼저 읽어 열 이름과 타입을 확인한다.
2. 네 개 CSV를 반복문으로 읽고, 각 파일에 `quarter` 열을 붙인다.
3. `concat` 후 행 수가 880인지 확인한다.
4. 날짜형 변환 후 월별 매출을 확인한다.
5. Excel 기준표를 읽고 `sku` 중복이 없는지 확인한다.
6. 매출 표와 기준표를 병합한 뒤 행 수와 결측을 확인한다.
7. 이익 계산 열을 만든 뒤 손계산 가능한 앞 5행을 검증한다.
8. JSON 캠페인 로그를 읽고 같은 분기 기준으로 요약한다.
9. 매출 요약과 캠페인 요약을 합쳐 최종 리포트를 만든다.
10. 최종 리포트를 저장하고, 결론 셀에 숫자와 다음 행동을 작성한다.

## 제출 품질 기준

| 항목 | 좋은 제출 | 보완이 필요한 제출 |
|---|---|---|
| 파일 경로 | `DATA_BASE` 기준 상대 경로 | 개인 컴퓨터 절대경로 |
| 중간 확인 | 합친 뒤 행 수, 병합 뒤 결측 확인 | 실행만 하고 검증 출력 없음 |
| 변수명 | `sales`, `merged`, `final_report` 처럼 의미가 분명함 | `df1`, `df2`, `data` 만 반복 |
| 저장 | 요약 리포트를 별도 CSV로 저장 | 원본 파일을 덮어씀 |
| 결론 | 분기, 금액, 전환, 다음 행동 포함 | "좋다", "높다"처럼 기준이 없음 |

## 자주 나는 오류와 확인 방법

- `FileNotFoundError`: 파일명이 맞는지, `DATA_BASE` 가 `"./data"` 인지 확인한다.
- `KeyError: 'sku'`: 지금 다루는 표에 `sku` 열이 있는지 `columns` 로 확인한다.
- `ImportError: openpyxl`: Excel 파일을 읽기 위한 패키지가 없는 환경이다. Colab이나 로컬 환경에서 `openpyxl` 설치가 필요할 수 있다.
- 병합 후 행 수 증가: 기준표의 `sku` 가 중복인지 확인한다.
- 병합 후 결측 발생: 매출 파일의 SKU가 기준표에 없는지 확인한다.
- 저장 파일이 안 보임: `os.path.exists(OUTPUT_PATH)` 로 현재 작업 폴더 기준 저장 여부를 확인한다.

## 결론 작성 가이드

최종 결론은 다음 네 요소를 포함한다.

- 가장 매출이 높은 분기와 금액
- 가장 매출총이익이 높은 분기와 금액
- 캠페인 전환 또는 이익/전환 효율이 좋은 분기
- 다음에 더 확인할 분석 방향

예를 들어 "4분기가 좋다"는 부족하다. "4분기는 매출과 매출총이익이 모두 1위였으므로, 다음에는 4분기 매출을 만든 SKU와 캠페인 채널을 같이 확인하겠다"처럼 기준과 행동이 같이 있어야 한다.
