# 파이썬 데이터 분석 커리큘럼

이 과정은 파이썬 기초 문법을 마친 학생이 실제 데이터를 다루며 분석 흐름을 익히는 10주 과정입니다. Google Colab에서 노트북을 실행하고, NumPy, pandas, matplotlib, seaborn을 사용해 데이터를 불러오고 정리하고 시각화하는 연습을 합니다.

## 과정 목표

| 목표 | 내용 |
|------|------|
| 데이터 이해 | 표 형태 데이터의 구조, 컬럼, 행, 타입을 읽고 설명합니다. |
| 분석 도구 활용 | NumPy와 pandas로 계산, 필터링, 정렬, 집계 작업을 수행합니다. |
| 데이터 정제 | 결측치, 중복, 이상치, 타입 오류를 찾아 수정합니다. |
| 시각화 | matplotlib과 seaborn으로 그래프를 만들고 패턴을 해석합니다. |
| 발표 정리 | 분석 결과를 근거와 함께 짧은 인사이트로 정리합니다. |

## 수업 방식

- 각 주차는 강의 노트북과 학생 실습 노트북으로 진행합니다.
- 학생 실습은 주차별 15문제를 기준으로 구성되어 있습니다.
- 문제는 완성 코드를 그대로 따라 치는 방식이 아니라, 빈칸과 힌트를 바탕으로 직접 완성하는 방식입니다.
- 수업 중에는 핵심 문제를 함께 풀고, 남은 문제는 복습이나 과제로 이어갈 수 있습니다.
- 마지막 주차에는 지금까지 배운 내용을 종합해 하나의 데이터 분석 흐름을 완성합니다.

## 주차별 학습 포인트

### 1강: 분석 개요 + NumPy 입문
Colab 환경에 익숙해지고, 데이터 분석이 어떤 순서로 진행되는지 배웁니다. NumPy 배열을 만들고 평균, 합계, 최댓값 같은 기본 통계를 계산합니다.

### 2강: NumPy 심화
2차원 배열을 다루며 `axis` 개념을 익힙니다. 브로드캐스팅과 표준화를 통해 여러 값을 한 번에 계산하는 방식을 연습합니다.

### 3강: pandas 기초
DataFrame의 행과 열을 이해하고, 필요한 컬럼을 선택하거나 조건에 맞는 행을 필터링합니다. 정렬과 빈도 계산으로 데이터를 빠르게 살펴봅니다.

### 4강: 데이터 불러오기와 저장
CSV, Excel, JSON 파일을 읽고 저장합니다. 여러 데이터를 합치거나 기준 컬럼으로 연결하는 방법을 배웁니다.

### 5강: 데이터 정제
실제 데이터에서 자주 만나는 결측치, 중복, 이상치, 타입 오류를 처리합니다. 분석 전에 데이터를 믿을 수 있는 상태로 만드는 과정입니다.

### 6강: 필터·정렬·파생 컬럼
조건에 맞는 데이터만 골라내고, 정렬 기준을 세우며, 새로운 컬럼을 만들어 분석에 필요한 feature를 구성합니다.

### 7강: 그룹화·집계·피벗
반, 과목, 카테고리처럼 그룹별로 평균과 합계를 계산합니다. 피벗 테이블을 이용해 한눈에 비교 가능한 요약표를 만듭니다.

### 8강: 시각화 기초
그래프를 통해 데이터의 차이와 흐름을 확인합니다. 그래프를 단순히 그리는 것에서 끝내지 않고, 무엇을 말해주는지 해석하는 연습을 합니다.

### 9강: 시계열 분석
날짜 데이터를 다루며 일별, 주별, 월별로 데이터를 묶어 봅니다. 이동평균과 변화량을 통해 흐름을 읽습니다.

### 10강: 통합 프로젝트
앞에서 배운 내용을 연결해 하나의 분석 과정을 완성합니다. 데이터 확인, 정제, 분석, 시각화, 인사이트 정리까지 전체 흐름을 경험합니다.
