### 2강 — NumPy 심화: 다차원 배열·브로드캐스팅·시뮬레이션

**강의 시간 배분 (총 120분)**

| 시간 | 내용 |
|------|------|
| 00~10분 | 1강 복습 퀴즈: "scores.mean() 와 scores.mean(axis=0) 차이?" |
| 10~30분 | reshape + ravel + flatten 라이브 코딩 |
| 30~50분 | axis 개념 시각화: 행 방향(0) vs 열 방향(1) |
| 50~70분 | 브로드캐스팅 규칙 설명 + 실습 |
| 70~95분 | 학생 실습: 문제 1~15 중 핵심 문제 진행 |
| 95~115분 | 표준화(Z-score) 해설 + 실무 활용 사례 |
| 115~120분 | 다음 주 pandas 예고 |

**강조 포인트**

- reshape의 `-1`: "나머지는 NumPy가 알아서 맞춰줘" — 가장 실용적인 패턴
- axis=0 vs axis=1: "0은 행을 따라 아래로 계산, 1은 열을 따라 옆으로 계산"
  - 시각적으로: 2D 배열을 그리고 화살표로 axis 방향 표시
- 브로드캐스팅: shape이 뒤에서부터 맞아야 함 (trailing dimensions)
  - `(3,4)` + `(4,)` → OK / `(3,4)` + `(3,)` → Error
- 표준화(Z-score): 평균 0, 표준편차 1로 만드는 것. ML에서 필수.

**학생이 자주 막히는 곳**

| 증상 | 원인 | 대응 |
|------|------|------|
| reshape 에러 | 원소 수가 맞지 않음 | `12`개짜리는 `(3,4)`, `(4,3)`, `(2,6)` 가능 |
| axis 혼동 | axis=0이 행인지 열인지 헷갈림 | "0번 차원을 따라 줄인다" = 행들을 합쳐서 하나 |
| 브로드캐스팅 에러 | shape 불일치 | `b.reshape(-1,1)` 로 열벡터 만들기 |

**문제 해설 포인트**
- 문제 5 (표준화): `z.mean(axis=0) ≈ 0`, `z.std(axis=0) ≈ 1` 확인으로 검증 습관 강조
- 문제 7 (np.where): pandas의 `np.where`와 `apply(lambda)`의 차이 언급

### 6강 — 필터·정렬·파생 컬럼: 성적표 feature engineering

**강의 시간 배분 (총 120분)**

| 시간 | 내용 |
|------|------|
| 00~15분 | feature engineering이란? ML에서의 중요성 |
| 15~40분 | assign으로 파생 컬럼 만들기 |
| 40~65분 | apply + 사용자 정의 함수 |
| 65~90분 | 학생 실습: 문제 1~15 중 핵심 문제 진행 |
| 90~110분 | 풀이 해설 (risk_label 함수 구조 설명) |
| 110~120분 | 다음 주 groupby 예고 |

**강조 포인트**

- `df.assign()`: 원본 불변 + 체이닝(메서드 연결) 가능 — 코드 가독성 ↑
- `apply(axis=1)`: 행 단위 함수 적용. 복잡한 조건 로직에 사용
- `lambda` vs 일반 함수: lambda는 단순 표현식, 복잡하면 def 함수
- 성능 주의: `apply`는 느림. 가능하면 벡터 연산(np.where, pd.cut) 선호

**risk_label 함수 설명 방법**

````
# 판서/슬라이드용 의사코드
def risk_label(row):
    if 출석률 < 75% 또는 과제 완료율 < 50%:
        return '위험'        # 즉시 개입 필요
    elif 평균 < 60:
        return '관찰'        # 모니터링 필요
    else:
        return '안전'        # 정상
````

| 시간 | 내용 |
|------|------|
| 00~15분 | 시계열 데이터란? 일반 테이블과 다른 점 |
| 15~40분 | datetime 인덱스 설정: `pd.to_datetime` + `set_index` |
| 40~65분 | resample + rolling 라이브 코딩 |
| 65~90분 | 학생 실습: 문제 1~15 중 핵심 문제 진행 |
| 90~110분 | shift + diff + pct_change 해설 |
| 110~120분 | 다음 주 최종 프로젝트 안내 |

**강조 포인트**

- `resample('ME')` 에서 'ME' = Month End. 'MS'=Month Start, 'W'=Weekly, 'D'=Daily
- rolling과 resample의 차이:
  - resample: 기간 단위로 데이터를 묶어 집계 (월별 합계 → 데이터 개수 줄어듦)
  - rolling: 이동 창으로 계산 (7일 이동평균 → 데이터 개수 그대로)
- `shift(1)`: "어제 값을 오늘 옆에" — 전일 대비 계산의 기본
- NaN 처리: rolling의 첫 (window-1)개, shift의 첫 n개는 NaN

**현업 활용 사례**

| 상황 | 사용 함수 |
|------|----------|
| 월별 매출 집계 | `resample('ME').sum()` |
| 7일 이동평균 대시보드 | `rolling(7).mean()` |
| 전주 대비 성장률 | `pct_change(7)` |
| 전일 대비 변화량 | `diff()` |

## 10. 강사 FAQ

**Q. 학생이 Colab에서 데이터 파일을 못 불러옵니다.**  
A. 노트북의 첫 번째 셀(`IS_COLAB` 설정 셀)을 먼저 실행했는지 확인. 실행 안 했으면 `DATA_BASE`가 `"./data"`로 설정되어 로컬 경로를 찾으려다 실패.

**Q. 학생이 셀을 순서 없이 실행해서 오류가 납니다.**  
A. Runtime → Restart and run all로 처음부터 순서대로 재실행. 이후에 셀 실행 순서가 중요하다는 점을 강조.

**Q. 학생이 빈칸에 아무것도 못 쓰고 막혀있습니다.**  
A. 힌트 주석을 다시 읽도록 유도. 힌트는 정답 함수명을 바로 주지 않고 빈칸 구조만 제공한다. 그래도 막히면 강의 노트의 같은 개념 예제를 찾아보게 한다.

**Q. 강의 시간이 부족합니다.**  
A. 문제 풀이는 숙제(`mission.ipynb`)로 넘기고 핵심 개념 설명에 집중. 빈칸 문제 일부는 다음 시간 시작 시 복습으로 활용.

**Q. 특정 학생이 진도가 너무 빠릅니다.**  
A. `final-mission.ipynb`로 심화 도전 안내. 또는 다른 학생 도우미 역할 부여.

**Q. 코드 에러를 학생이 읽지 못합니다.**  
A. 에러 메시지 읽기 훈련이 필수. "마지막 줄만 읽어라 (TypeError, ValueError 등)" → "몇 번째 줄인지 확인" → "그 줄 코드 집중 검토" 3단계 습관화.
