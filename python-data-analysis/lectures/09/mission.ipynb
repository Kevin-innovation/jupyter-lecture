| 문제 4~5 | 월별 집계와 주별 집계는 어떤 질문에 답하는가? |
| 문제 6~7 | 7일 이동평균과 14일 이동평균의 차이를 설명할 수 있는가? |
| 문제 8~10 | 변화량과 변화율은 언제 다르게 해석해야 하는가? |
| 문제 11~15 | 요일, 지연 상관, 이상치, 누적 지표를 결론에 어떻게 연결했는가? |

## 제출 형태 안내

- 날짜 범위와 집계 기준을 결론에서 명확하게 쓴다.
- `resample("ME")`, `resample("W")` 의 기준을 혼동하지 않는다.
- 이동평균은 원본 값을 대체하는 값이 아니라 추세를 보기 위한 보조 지표다.
- 이상치 후보는 오류라고 단정하지 말고 추가 확인 대상으로 표현한다.

## 풀이 전 작업 순서

1. 원본 CSV를 읽고 날짜 열 타입과 결측 여부를 확인한다.
2. `date` 를 `datetime` 으로 바꾸고 날짜 순서로 정렬한다.
3. 방문자, 주문, 매출, 광고비의 기본 파생 지표를 만든다.
4. 월별과 주별 리샘플을 먼저 만든다.
5. 7일, 14일 이동평균으로 일별 변동을 부드럽게 본다.
6. `shift` 와 `pct_change` 로 변화량과 변화율을 구분한다.
7. 요일별 패턴과 광고비 지연 상관은 보조 해석으로 사용한다.
8. 마지막에는 월별 흐름, 이동평균, 변화율, 이상치 후보를 함께 정리한다.

## 자주 나는 오류와 확인 방법

- `TypeError`: 날짜 열이 문자열인지 `df.dtypes` 를 확인한다.
- 월별 집계가 이상함: 날짜 인덱스를 설정했는지 확인한다.
- 첫 행 변화량이 비어 있음: 전날이 없으므로 정상이다.
- 이동평균 앞부분이 비어 있음: 창 크기만큼 데이터가 필요해서 정상이다.
- 전환율 평균이 이상함: 일별 평균이 아니라 합계 기준 전환율을 써야 하는 상황인지 확인한다.
- 결론에 "증가했다"라고 쓰기 전 실제 변화율이 양수인지 확인한다.

## 좋은 결론 조건

좋은 시계열 결론은 어느 시점에 변화가 있었는지와 그 변화가 어느 정도였는지를 함께 말한다.

- 매출이 가장 높았던 월과 날짜
- 이동평균 기준으로 흐름이 올라간 구간
- 전월 대비 변화율이 큰 월
- 이상치 후보와 추가 확인 행동

예를 들어 "3월 매출이 높다"는 부족하다. "3월 매출이 가장 높고, 2월 대비 증가율도 커서 캠페인 또는 외부 이벤트 여부를 함께 확인해야 한다"처럼 시점, 지표, 다음 행동이 연결되어야 한다.

## 시계열 해석 체크리스트

시계열 분석은 날짜 기준을 조금만 잘못 잡아도 실행은 되지만 결론이 틀어질 수 있다. 제출 전에는 아래 항목을 순서대로 확인한다.

| 항목 | 확인할 내용 |
|---|---|
| 날짜형 | `date` 가 문자열이 아니라 datetime 인가? |
| 정렬 | 날짜가 오름차순으로 정렬되어 있는가? |
| 리샘플 | 월별·주별 집계에서 합계와 평균을 구분했는가? |
| 이동평균 | 앞부분 NaN 을 오류로 지우지 않았는가? |
| 변화량 | 금액 차이와 퍼센트 변화율을 구분했는가? |
| 이상치 | 높은 값과 낮은 값을 바로 삭제하지 않고 후보로 표현했는가? |

월별 집계는 큰 흐름을 보여주고, 주별 집계는 짧은 변화 구간을 보여준다. 이동평균은 원본 값을 대체하는 것이 아니라 노이즈를 줄여 흐름을 보는 보조 지표다. 결론을 쓸 때는 "언제", "얼마나", "다음에 무엇을 확인할지"를 함께 적는다.

## 발표용 문장 작성 규칙

시계열 발표 문장은 날짜가 빠지면 설득력이 약해진다. "매출이 높았다"보다 "2025년 3월 매출이 가장 높았다"가 낫고, "증가했다"보다 "전월 대비 증가율이 가장 컸다"가 더 정확하다.

좋은 문장은 아래 네 가지를 포함한다.

1. 기간: 월, 주, 날짜 중 하나를 명확히 쓴다.
2. 지표: 방문자, 주문, 매출, 광고비 중 어떤 값을 말하는지 쓴다.
3. 기준: 최고값, 최저값, 증가율, 이동평균 중 무엇인지 쓴다.
4. 행동: 캠페인 확인, 휴일 확인, 광고비 조정처럼 다음 단계를 쓴다.
