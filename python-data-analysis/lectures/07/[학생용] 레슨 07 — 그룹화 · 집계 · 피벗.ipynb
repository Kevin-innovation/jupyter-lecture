- [ ] 문제 1~15 번호가 모두 남아 있다.
- [ ] 두 로그의 grain 차이를 설명할 수 있다.
- [ ] 매출 로그와 활동 로그를 바로 행 단위로 merge 하지 않았다.
- [ ] `groupby`, `pivot_table`, `crosstab` 을 모두 사용했다.
- [ ] 월별 결합 리포트가 코드 출력과 결론에서 같은 숫자를 사용한다.

## 질문 준비 메모

| 구간 | 스스로 확인할 질문 |
|---|---|
| 문제 1~2 | 두 로그의 한 행은 각각 무엇을 의미하는가? |
| 문제 3~6 | 매출을 카테고리, 채널, 월로 볼 때 질문이 어떻게 달라지는가? |
| 문제 7~8 | `pivot_table` 과 `crosstab` 은 어떤 상황에서 쓰는가? |
| 문제 9~13 | 이벤트 수, 사용자 수, 활동 시간은 서로 어떤 차이가 있는가? |
| 문제 14 | grain이 다른 데이터를 결합하기 전에 왜 월별 요약이 필요한가? |
| 문제 15 | 결론에 매출 규모와 활동 효율을 모두 포함했는가? |

## 제출 형태 안내

- 집계 기준을 변수명에 드러낸다. 예: `category_sales`, `monthly_activity`, `monthly_report`.
- 피벗 테이블은 행/열/값이 무엇인지 출력 전에 주석으로 남긴다.
- 결론에는 월, 카테고리, 액션, 효율 지표를 최소 하나씩 포함한다.
- 수치가 다르게 나오면 먼저 날짜형 변환과 월 컬럼 생성 여부를 확인한다.

## 풀이 전 작업 순서

1. 두 로그를 읽고 각 로그의 한 행이 무엇을 의미하는지 먼저 적는다.
2. 날짜형 변환 후 `month`, `weekday` 열을 만든다.
3. 매출 로그는 전체 매출, 수량, 평균 주문 매출부터 확인한다.
4. 카테고리, 채널, 지역, 월 순서로 매출을 점점 다른 기준에서 집계한다.
5. 피벗 테이블을 만들 때는 행, 열, 값, 집계 함수를 먼저 정한다.
6. 교차표를 만들 때는 빈도표인지 비율표인지 구분한다.
7. 활동 로그는 이벤트 수와 고유 사용자 수를 반드시 구분한다.
8. 액션별, 등급별, 사용자별 활동 요약을 만든다.
9. 두 로그를 결합할 때는 원본 행끼리 합치지 말고 월별 요약표끼리 합친다.
10. 결론에는 규모 지표와 효율 지표를 모두 포함한다.

## 자주 나는 오류와 확인 방법

- 원본 로그를 바로 `merge`: 두 로그의 grain이 달라서 잘못된 결합이다. 월별 요약 후 병합한다.
- `count` 와 `nunique` 혼동: 이벤트 수는 `size`, 사용자 수는 `nunique` 로 구한다.
- 월별 변화량이 이상함: `month` 기준 정렬 후 `diff()` 를 계산했는지 확인한다.
- 피벗 값이 원본과 다름: 피벗 행 합계와 원본 groupby 합계를 비교한다.
- 비율이 0~1인지 0~100인지 헷갈림: 퍼센트로 보여줄 때는 100을 곱하고 라벨에 `%` 를 적는다.
- 결론에 활동 로그가 없음: 7강은 매출과 활동을 연결하는 수업이므로 활동 지표를 반드시 포함한다.

## 좋은 결론 조건

최종 결론은 아래 네 축을 포함한다.

- 매출이 가장 높은 월과 매출액
- 활동 시간이 가장 긴 월 또는 이벤트가 가장 많은 액션
- 활동 대비 매출 효율이 가장 좋은 월
- 다음에 더 확인할 카테고리, 채널, 사용자 행동

예를 들어 "1월 매출이 높다"만 쓰면 부족하다. "1월은 매출과 활동 대비 매출 효율이 모두 높으므로, 1월의 gadget 매출과 review 활동이 어떤 채널에서 연결되는지 추가 확인하겠다"처럼 매출과 활동을 같이 해석해야 한다.

## 피벗과 교차표 작성 전 체크

피벗이나 교차표를 만들기 전에 아래 네 가지를 먼저 적어 둔다.

| 확인 항목 | 예시 |
|---|---|
| 행 기준 | `category`, `store_region`, `user_grade` |
| 열 기준 | `channel`, `action` |
| 값 | `revenue`, `minutes`, 또는 빈도 |
| 집계 방식 | `sum`, `mean`, `size` |

값 열이 있고 합계나 평균을 내고 싶으면 `pivot_table` 이 자연스럽다. 값 열 없이 조합별 개수만 세고 싶으면 `crosstab` 이 더 간단하다. 이 구분을 먼저 해 두면 7강 후반 문제에서 코드를 훨씬 안정적으로 작성할 수 있다.
