## 단계 1. 환경과 프로젝트 데이터

```python
import os
import pandas as pd
import numpy as np

IS_COLAB = "COLAB_GPU" in os.environ or "COLAB_TPU_ADDR" in os.environ
if IS_COLAB:
    DATA_BASE = "https://raw.githubusercontent.com/Kevin-innovation/jupyter-lecture/main/python-data-analysis/lectures/10/data"
else:
    DATA_BASE = "./data"
print("data base:", DATA_BASE)

import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
df = pd.read_csv(f'{DATA_BASE}/city_learning_programs.csv')
print(df.shape)
print(df.head())
print(df.info())
```

## 단계 2. 문제 정의와 KPI

```python
df['start_date'] = pd.to_datetime(df['start_date'])
df['cost_per_participant'] = df['budget'] / df['participants']
df['effective_completions'] = df['participants'] * df['completion_rate']
print(df[['participants','budget','completion_rate','satisfaction','cost_per_participant']].describe())
```

## 단계 3. 그룹별 성과

```python
program_summary = df.groupby('program_type').agg(programs=('program_id','count'), participants=('participants','sum'), completion_rate=('completion_rate','mean'), satisfaction=('satisfaction','mean'), budget=('budget','sum'))
print(program_summary.sort_values('participants', ascending=False))
district_pivot = pd.pivot_table(df, values='participants', index='district', columns='program_type', aggfunc='sum', fill_value=0)
print(district_pivot)
```

## 단계 4. 시각화와 결론 후보

```python
fig, ax = plt.subplots(figsize=(6,3))
sns.barplot(data=df, x='program_type', y='participants', estimator='sum', errorbar=None, ax=ax)
plt.close(fig)
fig, ax = plt.subplots(figsize=(5,3))
sns.scatterplot(data=df, x='cost_per_participant', y='satisfaction', hue='program_type', ax=ax)
plt.close(fig)
print('project visuals ready')
```

## 답안 변형 허용 범위

학생이 모범 답안과 다른 변수명을 써도 괜찮다. 다른 함수 조합으로 같은 표를 만들 수도 있다. 허용 가능한 변형은 다음과 같다.

- 중간 표 이름이 다르지만 행과 열 기준이 같음.
- 출력 형식을 원 단위, 만원 단위, 퍼센트 단위 등으로 바꾸었지만 단위가 명시됨.
- 결론 문장 순서가 다르지만 핵심 숫자와 추천 방향이 일치함.
- 보너스에서 다른 비교 기준을 사용했지만 데이터 근거가 있음.

허용하지 않는 변형은 다음과 같다.

- 절대경로를 사용해 다른 환경에서 실행되지 않음.
- 중간 표가 비어 있는데도 결론을 작성함.
- 출력과 반대되는 결론을 씀.
- 학생용 미션 파일에 정답 코드를 그대로 붙여넣음.

## 강사용 구두 질문

네 질문에 답할 수 있으면 코드 복붙이 아니라 분석 흐름을 이해한 것으로 볼 수 있다.

## 재제출을 요청해야 하는 경우

- 핵심 결과 표가 없고 원본 전체만 출력했다.
- 결론이 비어 있거나 “잘 모르겠다” 수준으로 끝난다.
- 다른 학생 답안을 그대로 복사한 흔적이 있다.

재제출을 요청할 때는 모든 것을 다시 하게 하기보다 어느 단계부터 고치면 되는지 짚어 준다. 대부분은 데이터 로드, 중간 표 기준, 결론 문장 중 하나만 고치면 된다.
