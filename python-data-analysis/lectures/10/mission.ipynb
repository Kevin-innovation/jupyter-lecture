- [ ] 문제 1~15 번호가 모두 남아 있다.
- [ ] 날짜, 월, 분기, 비용, 완료자 수, 효율 점수를 만들었다.
- [ ] 프로그램 유형, 지역, 대상 그룹 기준 요약을 모두 만들었다.
- [ ] 피벗, 산점도, 선 그래프 중 최소 2개 이상을 사용했다.
- [ ] 최종 결론이 코드 출력 숫자와 모순되지 않는다.

## 질문 준비 메모

| 구간 | 스스로 확인할 질문 |
|---|---|
| 문제 1~4 | 데이터 품질과 핵심 KPI를 어떻게 확인했는가? |
| 문제 5~8 | 프로그램 유형, 지역, 대상 그룹, 피벗은 각각 어떤 질문에 답하는가? |

## 제출 형태 안내

- 모든 결론에는 숫자, 비교 대상, 다음 행동을 포함한다.
- 효율 점수는 절대 진리가 아니라 운영 우선순위를 보기 위한 가중치 예시로 설명한다.
- 좋은 프로그램과 점검 대상 프로그램의 조건을 코드와 글에서 같은 기준으로 쓴다.
- 발표용 문장은 긴 문단보다 짧고 명확한 5개 문장으로 정리한다.

## 풀이 전 작업 순서

1. 데이터의 행 수, 열 이름, 결측, 중복 ID를 확인한다.
2. 시작일을 날짜형으로 바꾸고 월·분기 기준을 만든다.
3. 참여자 수, 예산, 완료율, 만족도 범위를 점검한다.
4. 1인당 비용, 예상 완료자 수, 완료율 구간을 만든다.
5. 프로그램 유형, 지역, 대상 그룹 기준으로 요약한다.
6. 피벗 테이블로 지역과 프로그램 유형 조합을 본다.
7. 우수 프로그램과 점검 대상 프로그램 기준을 정한다.
8. 효율 점수와 차트로 발표용 근거를 만든다.
9. 마지막에는 전체 규모, 강점, 위험, 다음 행동을 짧게 쓴다.

## 자주 나는 오류와 확인 방법

- `KeyError`: 열 이름 오타가 있는지 `df.columns` 를 확인한다.
- 날짜 변환 오류: `start_date` 형식을 먼저 확인한다.
- 비용 점수 방향 오류: 1인당 비용이 낮을수록 높은 점수인지 확인한다.
- 완료율을 0~100으로 착각함: 원본 완료율은 0~1 사이 값이다.
- 피벗 합계가 원본 합계와 다름: `fill_value=0` 과 집계 기준을 확인한다.
- 결론이 너무 일반적임: 숫자와 다음 행동이 들어갔는지 확인한다.

## 좋은 결론 조건

좋은 프로젝트 결론은 "어떤 프로그램이 좋다"에서 끝나지 않는다. 왜 좋은지, 어떤 기준에서 봤는지, 다음에 무엇을 해야 하는지를 함께 말한다.

- 전체 운영 규모와 핵심 KPI
- 참여자 수가 많은 프로그램 유형 또는 지역
- 만족도와 완료율이 높은 우수 그룹
- 점검 대상 프로그램 수와 이유
- 다음 운영 개선 행동

예를 들어 "math 프로그램이 좋다"는 부족하다. "math 프로그램은 참여자 수가 많지만 1인당 비용과 완료율을 함께 보면 우선순위가 달라질 수 있으므로, 다음 운영 회의에서는 참여자 수와 효율 점수를 함께 비교해야 한다"처럼 기준 차이를 설명해야 한다.

## 프로젝트 제출 체크리스트

최종 프로젝트는 코드 셀만 채워서는 완성되지 않는다. 제출 전에 아래 항목이 모두 들어갔는지 확인한다.

| 항목 | 포함할 내용 |
|---|---|
| 데이터 품질 | 결측, 중복 ID, 값 범위 |
| 규모 | 프로그램 수, 참여자 수, 예산 |
| 품질 | 완료율, 만족도, 우수 프로그램 수 |
| 효율 | 1인당 비용, 효율 점수 |
| 비교 | 프로그램 유형, 지역, 대상 그룹 |
| 행동 | 확대할 기준과 점검할 기준 |

프로젝트 발표에서는 모든 표를 보여줄 필요가 없다. 전체 대시보드 1개, 핵심 비교표 1개, 차트 2개, 최종 결론 5문장이면 충분하다. 중요한 것은 표와 결론이 서로 같은 숫자를 말하는 것이다.
