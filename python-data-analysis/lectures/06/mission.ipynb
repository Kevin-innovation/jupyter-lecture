## 풀이 전 작업 순서

1. 원본 성적표를 읽고 행 수, 열 이름, 결측 여부를 확인한다.
2. 점수 열 세 개를 `score_cols` 로 묶어 둔다.
3. 학생별 평균과 총점을 먼저 만든다.
4. 우수 학생, 낮은 평균 학생, 낮은 출석률, 낮은 과제 제출률을 각각 따로 필터링한다.
5. 복합 조건을 만들 때는 각 조건을 괄호로 감싼다.
6. 학년별·반별 groupby 요약으로 전체 흐름을 본다.
7. 운영용 핵심 점수처럼 가중치를 쓰는 지표는 계산식을 주석으로 남긴다.
8. 라벨과 순위를 만든 뒤 상태별 인원 합이 전체 학생 수와 같은지 확인한다.
9. `apply` 로 만든 피드백 문장은 앞 10개 이상을 출력해 문장이 어색하지 않은지 본다.

## 자주 나는 오류와 확인 방법

- `KeyError`: 열 이름 오타가 있는지 `df.columns` 를 확인한다.
- 평균이 과목별로 나옴: `axis=1` 을 써야 학생별 평균이 나온다.
- 복합 조건 오류: `and`, `or` 대신 `&`, `|` 를 사용하고 조건마다 괄호를 붙인다.
- 핵심 점수가 너무 낮음: 출석률과 과제 제출률에 100을 곱했는지 확인한다.
- 반별 순위가 전체 순위와 같음: `groupby("class_name")` 뒤에 `rank` 를 적용했는지 확인한다.
- 상태별 합계가 480명이 아님: `student_status` 결측이나 조건 누락이 있는지 확인한다.

## 좋은 결론 조건

좋은 결론은 단순히 1등을 말하는 데서 끝나지 않는다. 아래 네 요소가 들어가야 한다.

- 평균 점수 1위와 핵심 점수 1위 비교
- 지원 후보 수와 우수 후보 수
- 관리가 필요한 학년 또는 반
- 다음 수업이나 운영에서 확인할 행동

예를 들어 "S0439가 1등이다"는 부족하다. "평균 점수 1위는 S0439이지만 핵심 점수 1위는 출석과 과제까지 높은 S0110이므로, 운영 지표에서는 학습 습관을 함께 봐야 한다"처럼 기준 차이를 설명해야 한다.
