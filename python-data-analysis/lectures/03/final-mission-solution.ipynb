## 1. 환경과 데이터 로드

```python
import os
import pandas as pd

IS_COLAB = "COLAB_GPU" in os.environ or "COLAB_TPU_ADDR" in os.environ
if IS_COLAB:
    DATA_BASE = "https://raw.githubusercontent.com/Kevin-innovation/jupyter-lecture/main/python-data-analysis/lectures/03/data"
else:
    DATA_BASE = "./data"
print("data base:", DATA_BASE)

df = pd.read_csv(f"{DATA_BASE}/sales_orders.csv")
print("shape:", df.shape)
print("columns:", list(df.columns))
print(df.info())
```

## 2. 매출 열 만들기

```python
df["gross_revenue"] = df["quantity"] * df["unit_price"]
df["net_revenue"] = df["gross_revenue"] * (1 - df["discount_rate"])
print(df[["quantity", "unit_price", "discount_rate", "net_revenue"]].head())
```

## 3. 전체 KPI

```python
total_revenue = df["net_revenue"].sum()
avg_order = df["net_revenue"].mean()
median_order = df["net_revenue"].median()
order_count = len(df)

print("주문 수:", order_count)
print("총 순매출:", f"{total_revenue:,.0f}원")
print("평균 주문금액:", f"{avg_order:,.0f}원")
print("중앙 주문금액:", f"{median_order:,.0f}원")
```

## 4. 채널과 카테고리 분포

```python
channel_counts = df["channel"].value_counts()
channel_ratio = df["channel"].value_counts(normalize=True) * 100
category_counts = df["category"].value_counts()

print("채널 주문 수:\n", channel_counts)
print("채널 비율(%):\n", channel_ratio.round(1))
print("카테고리 주문 수:\n", category_counts)

best_channel = channel_counts.idxmax()
best_category = category_counts.idxmax()
```

## 5. 상위 주문 표

```python
top10 = df.sort_values("net_revenue", ascending=False).head(10)
print(top10[["order_id", "date", "channel", "category", "quantity", "unit_price", "discount_rate", "net_revenue"]])
```

## 6. 보너스 계산

```python
region_counts = df["region"].value_counts()
print("지역별 주문 수:\n", region_counts)
print("주문 수 1위 지역:", region_counts.idxmax())
```

```python
vip = df[df["customer_grade"] == "V"]
print("VIP 평균 주문금액:", vip["net_revenue"].mean())
print("전체 평균 주문금액:", df["net_revenue"].mean())
```

```python
no_discount = df[df["discount_rate"] == 0]
with_discount = df[df["discount_rate"] > 0]
print("무할인 평균:", no_discount["net_revenue"].mean())
print("할인 평균:", with_discount["net_revenue"].mean())
```

## 교사용 해석 가이드

학생 결론은 숫자가 조금 다르게 표현되어도 괜찮다. 예를 들어 총 순매출을 원 단위로 쓰거나 만원 단위로 반올림해도 코드 출력과 일관되면 통과다. 중요한 것은 “전체 규모 → 어떤 축이 중요함 → 다음 분석”의 흐름이다.

좋은 답안은 보통 다음 특징을 가진다.

- `net_revenue` 를 만든 뒤 모든 KPI 를 같은 열에서 계산한다.
- `value_counts` 결과를 그대로 복사하지 않고 1위와 비율을 문장으로 바꾼다.
- 상위 주문 표를 보고 고가 상품 또는 높은 수량의 영향을 언급한다.
- 보너스에서는 VIP, 지역, 할인 여부 중 하나를 추가 분석 방향으로 연결한다.

2. 주문 수 1위 채널이 매출 1위 채널이라고 단정할 수 있는가?

## 오답 처리 기준

- `gross_revenue` 만 계산하고 `net_revenue` 를 만들지 않은 답안은 필수 요구사항 미달이다.
- 상위 주문 표가 오름차순이면 코드 한 줄 수정 후 재제출하게 한다.
- 결론에 “채널이 중요하다”처럼 대상명이 빠져 있으면 결론 보완을 요구한다.
- 보너스 계산은 필수 점수 부족을 일부 보완할 수 있지만, 필수 매출 계산 오류를 덮지는 못한다.
