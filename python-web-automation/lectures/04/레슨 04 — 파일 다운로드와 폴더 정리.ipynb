## 1. 파일 링크 구조

자료실 HTML의 a.file-link는 href와 data-type, data-week 속성을 가진다.


```python
import os
import re
import time
import csv
from pathlib import Path
from urllib.parse import urljoin, urlparse, parse_qs, urlencode

try:
    import requests
    from bs4 import BeautifulSoup
except ImportError:
    import sys, subprocess
    subprocess.check_call([sys.executable, '-m', 'pip', '-q', 'install', 'requests', 'beautifulsoup4'])
    import requests
    from bs4 import BeautifulSoup

IS_COLAB = 'COLAB_GPU' in os.environ or 'COLAB_TPU_ADDR' in os.environ
if IS_COLAB:
    DATA_BASE = 'https://raw.githubusercontent.com/Kevin-innovation/jupyter-lecture/main/python-web-automation/lectures/04/data'
else:
    DATA_BASE = './data'

def load_text(filename):
    if DATA_BASE.startswith('http'):
        url = f'{DATA_BASE}/{filename}'
        response = requests.get(url, timeout=10, headers={'User-Agent': 'D-Lab-Lesson/1.0'})
        response.raise_for_status()
        response.encoding = response.encoding or 'utf-8'
        return response.text
    return Path(DATA_BASE, filename).read_text(encoding='utf-8')

def load_bytes(filename):
    if DATA_BASE.startswith('http'):
        url = f'{DATA_BASE}/{filename}'
        response = requests.get(url, timeout=10, headers={'User-Agent': 'D-Lab-Lesson/1.0'})
        response.raise_for_status()
        return response.content
    return Path(DATA_BASE, filename).read_bytes()

def clean_int(text):
    return int(re.sub(r'[^0-9]', '', str(text)))

def safe_filename(name):
    return re.sub(r'[^A-Za-z0-9_.-]+', '_', str(name)).strip('_')

def save_bytes(relative_path, target_dir):
    target_dir = Path(target_dir)
    target_dir.mkdir(parents=True, exist_ok=True)
    filename = safe_filename(Path(relative_path).name)
    target = target_dir / filename
    target.write_bytes(load_bytes(relative_path))
    return target

print('colab:', IS_COLAB)
print('data base:', DATA_BASE)

html_text = load_text('resource_center.html')
soup = BeautifulSoup(html_text, 'html.parser')
links = soup.select('a.file-link')
print(len(links), links[0]['href'])
```

## 강의 보강 노트

이 절은 수업 중 교사가 질문으로 풀어낼 수 있는 운영형 설명이다. 학생이 셀을 실행한 뒤 결과만 맞히지 않고 자동화 절차를 말로 설명하도록 돕는다.

### 1. 다운로드 대상 식별

자료실 자동화는 링크를 모두 받는 것이 아니라 받을 파일과 건너뛸 파일을 구분하는 작업이다. 확장자, 파일 크기, 카테고리, 공개 상태를 기준으로 선택하게 하면 무분별한 다운로드를 막을 수 있다.

### 2. 폴더 구조 설계

저장 폴더를 먼저 정하지 않으면 결과 파일이 코랩 작업 폴더에 뒤섞인다. 날짜, 과정명, 파일 유형을 기준으로 폴더를 나누는 예시는 이후 운영 자동화에서도 바로 쓰인다.

### 3. 파일명 정리

웹에서 받은 파일명에는 공백, 괄호, 한글, 특수문자가 섞일 수 있다. 학생에게 원본 파일명을 보존하되 저장용 이름은 안전하게 정리하는 이유를 설명하게 한다.

### 4. 중복 다운로드 방지

같은 파일을 여러 번 받지 않으려면 URL, 파일명, 크기 중 어떤 값을 key로 삼을지 정해야 한다. fixture에서는 manifest를 기준으로 중복 판단을 연습하고, 실제 서비스에서는 수정일도 함께 고려한다.

### 5. 실패 기록

다운로드 실패를 단순히 print로 지나가면 나중에 누락을 찾기 어렵다. 실패한 URL, 이유, 재시도 여부를 별도 로그에 남기는 구조를 기본 답안으로 본다.

### 6. 자료실 안전 규칙

대용량 파일과 비공개 자료가 섞인 실제 사이트에서는 자동 다운로드가 쉽게 문제를 만든다. 수업에서는 합성 파일만 다루고, 실제 적용 전에는 권한과 저장 위치를 반드시 확인한다.

### 7. 검수 방식

다운로드 자동화의 완료 기준은 파일 개수만이 아니다. 파일이 열리는지, 크기가 0이 아닌지, manifest와 실제 저장 목록이 맞는지까지 확인해야 운영에 쓸 수 있다.

## 0. 환경 셀


```python
import os
import re
import time
import csv
from pathlib import Path
from urllib.parse import urljoin, urlparse, parse_qs, urlencode

try:
    import requests
    from bs4 import BeautifulSoup
except ImportError:
    import sys, subprocess
    subprocess.check_call([sys.executable, '-m', 'pip', '-q', 'install', 'requests', 'beautifulsoup4'])
    import requests
    from bs4 import BeautifulSoup

IS_COLAB = 'COLAB_GPU' in os.environ or 'COLAB_TPU_ADDR' in os.environ
if IS_COLAB:
    DATA_BASE = 'https://raw.githubusercontent.com/Kevin-innovation/jupyter-lecture/main/python-web-automation/lectures/04/data'
else:
    DATA_BASE = './data'

def load_text(filename):
    if DATA_BASE.startswith('http'):
        url = f'{DATA_BASE}/{filename}'
        response = requests.get(url, timeout=10, headers={'User-Agent': 'D-Lab-Lesson/1.0'})
        response.raise_for_status()
        response.encoding = response.encoding or 'utf-8'
        return response.text
    return Path(DATA_BASE, filename).read_text(encoding='utf-8')

def load_bytes(filename):
    if DATA_BASE.startswith('http'):
        url = f'{DATA_BASE}/{filename}'
        response = requests.get(url, timeout=10, headers={'User-Agent': 'D-Lab-Lesson/1.0'})
        response.raise_for_status()
        return response.content
    return Path(DATA_BASE, filename).read_bytes()

def clean_int(text):
    return int(re.sub(r'[^0-9]', '', str(text)))

def safe_filename(name):
    return re.sub(r'[^A-Za-z0-9_.-]+', '_', str(name)).strip('_')

def save_bytes(relative_path, target_dir):
    target_dir = Path(target_dir)
    target_dir.mkdir(parents=True, exist_ok=True)
    filename = safe_filename(Path(relative_path).name)
    target = target_dir / filename
    target.write_bytes(load_bytes(relative_path))
    return target

print('colab:', IS_COLAB)
print('data base:', DATA_BASE)
```

자료실 HTML과 manifest.csv를 기준으로 수업 자료를 그룹별 폴더에 저장하고 다운로드 로그를 만든다.

## 0. 환경 셀


```python
import os
import re
import time
import csv
from pathlib import Path
from urllib.parse import urljoin, urlparse, parse_qs, urlencode

try:
    import requests
    from bs4 import BeautifulSoup
except ImportError:
    import sys, subprocess
    subprocess.check_call([sys.executable, '-m', 'pip', '-q', 'install', 'requests', 'beautifulsoup4'])
    import requests
    from bs4 import BeautifulSoup

IS_COLAB = 'COLAB_GPU' in os.environ or 'COLAB_TPU_ADDR' in os.environ
if IS_COLAB:
    DATA_BASE = 'https://raw.githubusercontent.com/Kevin-innovation/jupyter-lecture/main/python-web-automation/lectures/04/data'
else:
    DATA_BASE = './data'

def load_text(filename):
    if DATA_BASE.startswith('http'):
        url = f'{DATA_BASE}/{filename}'
        response = requests.get(url, timeout=10, headers={'User-Agent': 'D-Lab-Lesson/1.0'})
        response.raise_for_status()
        response.encoding = response.encoding or 'utf-8'
        return response.text
    return Path(DATA_BASE, filename).read_text(encoding='utf-8')

def load_bytes(filename):
    if DATA_BASE.startswith('http'):
        url = f'{DATA_BASE}/{filename}'
        response = requests.get(url, timeout=10, headers={'User-Agent': 'D-Lab-Lesson/1.0'})
        response.raise_for_status()
        return response.content
    return Path(DATA_BASE, filename).read_bytes()

def clean_int(text):
    return int(re.sub(r'[^0-9]', '', str(text)))

def safe_filename(name):
    return re.sub(r'[^A-Za-z0-9_.-]+', '_', str(name)).strip('_')

def save_bytes(relative_path, target_dir):
    target_dir = Path(target_dir)
    target_dir.mkdir(parents=True, exist_ok=True)
    filename = safe_filename(Path(relative_path).name)
    target = target_dir / filename
    target.write_bytes(load_bytes(relative_path))
    return target

print('colab:', IS_COLAB)
print('data base:', DATA_BASE)
```


## 모범 답안


```python
manifest = list(csv.DictReader(load_text('manifest.csv').splitlines()))
logs = []
for row in manifest:
    path = save_bytes(row['path'], Path('downloads/lesson04/final') / row['group'])
    logs.append({'file_name': row['file_name'], 'group': row['group'], 'saved_path': str(path), 'size': path.stat().st_size})
with open('lesson04_download_log.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['file_name','group','saved_path','size'])
    writer.writeheader(); writer.writerows(logs)
print('logs:', len(logs))
print('total bytes:', sum(row['size'] for row in logs))
```
